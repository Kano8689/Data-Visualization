## **Data Cleaning**

#### **Rename Columns**

```python
df.columns = (df.columns.str.strip()
                        .str.lower()
                        .str.replace("!", "")
                        .str.replace("$", "")
                        .str.replace("(", "")
                        .str.replace(")", "")
                        .str.strip()
                        .str.replace("/", "_")
                        .str.replace("-", "_")
                        .str.replace(" ", "_")
                        .str.replace("_%", "_pct")
                        )

# df.columns.tolist()

# for i in range(len(df.columns.to_list())):
#       print(f"{i}: {df.columns.to_list()[i]}")

df = df.rename(columns={
    'date': 'date',
    'stock_index_type': 'stock_index',
    'open_price': 'open_price',
    'close_price_': 'close_price',
    'daily_high': 'daily_high',
    'daily_low': 'daily_low',
    'trading_volume_vol': 'trading_volume',
    'gdp_growth_rate_%': 'gdp_growth_pct',
    'inflation_rate_pct': 'inflation_pct',
    'unemployment_rate_%': 'unemployment_pct',
    'interest_rate_pct': 'interest_pct',
    'government_debt_billion_usd': 'gov_debt_billion',
    'corporate_profits_billion_usd': 'corporate_profits_billion',
    'forex_usd_eur': 'forex_usd_eur',
    'forex_usd_jpy': 'forex_usd_jpy',
    'crude_oil_price_per_barrel': 'crude_oil_bbl',
    'gold_price_usd_ounce': 'gold_price_oz',
    'retail_sales_billion_usd': 'retail_sales_billion',
    'bankruptcy_rate_pct': 'bankruptcy_pct',
    'mergers_&_acquisitions_deals': 'ma_deals',
    'venture_capital_funding_billion_usd': 'vc_funding_billion',
    'import_billion_usd': 'import_billion_usd',
    'export_billion_usd__': 'export_billion_usd',
})

for i in range(len(df.columns.to_list())):
      print(f"{i}: {df.columns.to_list()[i]}")
```

#### **Garbage Values Convert into *np.nan* Values**

```python
garbage_nulls = ['?', 'N/A', 'null', 'none', 'None', 'missing', 'nan', 'NAN', 'Nan', '-9999']
df = df.replace(garbage_nulls, np.nan)
df = df.applymap(lambda x: np.nan if isinstance(x, str) and x.strip() == '' else x)
```

#### **Replace Garbage Values with Numeric Values**

```python
text_numeric_cols = df.columns.to_list()
text_numeric_cols = text_numeric_cols[2:]
text_numeric_cols
```

```python
garbage_suffixes = ['usd', 'USD', '\\$', 'h', 'H', 'l', 'L', 'px', '%', '$', 'b', 'B', 'eur', 'jpy', '/bbl', '/oz', 'pts', 
                    'idx', 'deals' ]

for col in text_numeric_cols:
      if col in df.columns:
            df[col] = df[col].astype(str)
            for suffix in garbage_suffixes:
                  df[col] = df[col].str.strip().str.replace(suffix, "", case=False, regex=False).str.replace(" ","").str.strip()
            df[col] = df[col].str.replace(",", ".", regex=False)
            df[col] = pd.to_numeric(df[col].str.strip(), errors='coerce')

df.head(10)
```

```python
df.isnull().sum()
```

#### **Standardalize Text**

```python
month_map = {
      'Jan':'01','Feb':'02','Mar':'03','Apr':'04','May':'05','Jun':'06',
      'Jul':'07','Aug':'08','Sep':'09','Oct':'10','Nov':'11','Dec':'12',
}

date_list = date_mmm_list = list()

for i, val in enumerate(df["date"].astype(str)):
      parts = val.split('-')

      if len(parts) == 3:
            if parts[0] in list(month_map):
                  parts[0] = month_map[parts[0]]
                  val = f"{parts[1]}-{parts[0]}-{parts[2]}"

      df.at[i, "date"] = val

df['date'] = pd.to_datetime(df['date'], errors='coerce')
df = df.dropna(subset=['date'])
df = df.sort_values('date').reset_index(drop=True)
df.head()
```

```python
df['stock_index'] = df['stock_index'].astype(str).str.strip()

index_mapping = {
    'dow jones': 'Dow Jones', 'DOW': 'Dow Jones', 'DJIA': 'Dow Jones', 'Dow Jones ': 'Dow Jones',
    's&p 500': 'S&P 500', 'S&P500': 'S&P 500', 'S&P 500 Index': 'S&P 500', 'S&P 500 Index': 'S&P 500',
    'nasdaq': 'NASDAQ', 'Nasdaq': 'NASDAQ', 'NASDAQ ': 'NASDAQ'
}

df['stock_index'] = df['stock_index'].replace(index_mapping)
df.loc[df['stock_index'] == 'nan', 'stock_index'] = np.nan
df['stock_index'] = df['stock_index'].fillna("Unknown")
```

```python
interpolate_columns = ['open_price', 'close_price', 'daily_high', 'daily_low', 
                       'forex_usd_eur', 'forex_usd_jpy', 'crude_oil_bbl', 'gold_price_oz'
                       ]
ffill_columns = ['gdp_growth_rate_pct', 'inflation_pct', 'unemployment_rate_pct', 'interest_pct',
                 'consumer_confidence_index', 'gov_debt_billion', 'corporate_profits_billion',
                 'real_estate_index', 'retail_sales_billion', 'consumer_spending_billion_usd',
                 'import_billion_usd', 'export_billion_usd', 'bankruptcy_pct'
                 ]
median_columns = ['trading_volume', 'ma_deals', 'vc_funding_billion']
volume_cols = ['trading_volume', 'ma_deals', 'vc_funding_billion']

df[interpolate_columns] = df[interpolate_columns].interpolate(method='linear')
df[ffill_columns] = df[ffill_columns].ffill()

for col in volume_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.head()
```

```python
df.loc[df['trading_volume'] < 0, 'trading_volume'] = np.nan
df.loc[df['gdp_growth_rate_pct'] > 100, 'gdp_growth_rate_pct'] = np.nan
df.loc[df['interest_pct'] < -10, 'interest_pct'] = np.nan
```

```python
df = df.drop_duplicates()

df = df.drop_duplicates(subset=['date', 'stock_index'], keep='first')

numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```
