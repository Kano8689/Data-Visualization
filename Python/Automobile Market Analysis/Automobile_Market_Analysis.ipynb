```python
data.rename(columns={"Make":"Company","Engine Fuel Type":"Fuel Type", "highway MPG":"Highway MPG", "city mpg":"City MPG", "Engine HP":"HP", "Engine Cylinders":"Cylinders", "Driven_Wheels":"Driven Wheels"}, inplace=True)
data.head()
```

```python
data = data.drop_duplicates()
data.head(3)
```

```python
data.count()
```

```python
data1 = data.copy()
```

```python
missing_data = data1.isnull().sum().reset_index()
missing_data.columns = ["Column","Missing Values"]
fig = px.bar(missing_data, x="Column", y="Missing Values", title="Missing Values in Each Column", color="Missing Values", color_continuous_scale="Blues", text_auto=True)
fig.show()
```

```python
data["Highway MPG"].fillna(int(data["Highway MPG"].mean()), inplace=True)
data["City MPG"].fillna(int(data["City MPG"].mean()), inplace=True)
data["Cylinders"].fillna(int(data["Cylinders"].mode()), inplace=True)
data["Number of Doors"].fillna(int(data["Number of Doors"].mode()), inplace=True)
data["Year"].bfill(inplace=True)
data["Company"].fillna("Unknown", inplace=True)
data["Fuel Type"] = data["Fuel Type"].fillna(
    data.groupby("Company")["Fuel Type"].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else x)
)
data["Model"] = data["Model"].fillna(
    data.groupby("Company")["Model"].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else x)
)
data["Transmission Type"].bfill(inplace=True)
data["Driven Wheels"] = data['Driven Wheels'].fillna(
    data.groupby("Transmission Type")["Driven Wheels"].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else x)
)
data["Market Category"] = data["Market Category"].fillna(
    data.groupby(["Company","Number of Doors"])["Market Category"].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else x)
)
data["Vehicle Size"] = data["Vehicle Size"].fillna(
    data.groupby("Market Category")["Vehicle Size"].transform(
        lambda x: x.mode().iloc[0] if not x.mode().empty else "Unknown"
    )
)
data["Vehicle Style"] = data["Vehicle Size"].fillna(
    data.groupby(["Company","Market Category","Vehicle Size"])["Vehicle Style"].transform(lambda x: x.mode().iloc[0] if not x.mode().empty else x)
)
```

```python
sns.countplot(x='Company', hue="Model", data=data)
plt.title('Car Company Names')
plt.show()
```

```python
sns.countplot(x='Market Category', data=data)
plt.title('Market Category')
plt.show()
```

```python
sns.countplot(x='Transmission Type', hue="Driven Wheels",data=data)
plt.title('Transmission Type')
plt.show()
```

```python
sns.violinplot(x="Market Category", y="Popularity", hue="Transmission Type",data=data)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
company_counts = data['Company'].value_counts().nlargest(10).sort_values(ascending=False)
sns.barplot(x=company_counts.values, y=company_counts.index, palette='viridis')
plt.title('Top 10 Car Companies by Count', fontsize=14)
plt.xlabel('Number of Cars')
plt.ylabel('Company')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
fuel_msrp = data.groupby('Fuel Type')['MSRP'].mean().sort_values(ascending=False).reset_index()
sns.barplot(data=fuel_msrp, x='MSRP', y='Fuel Type', palette='magma')
plt.title('Average MSRP by Fuel Type', fontsize=14)
plt.xlabel('Average MSRP ($)')
plt.ylabel('Fuel Type')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='HP', y='MSRP', hue='Fuel Type', alpha=0.7, s=100)
plt.title('Relationship: Horsepower (HP) vs. MSRP', fontsize=14)
plt.xlabel('Horsepower')
plt.ylabel('MSRP ($)')
plt.legend(title='Fuel Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
style_counts = data['Vehicle Style'].value_counts().sort_values(ascending=False)
sns.barplot(x=style_counts.values, y=style_counts.index, palette='coolwarm')
plt.title('Distribution of Vehicle Body Styles', fontsize=14)
plt.xlabel('Count')
plt.ylabel('Vehicle Style')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
mpg_transmission = data.groupby('Transmission Type')[['Highway MPG', 'City MPG']].mean().reset_index()
mpg_melted = mpg_transmission.melt(id_vars='Transmission Type', var_name='MPG Type', value_name='Average MPG')

sns.barplot(data=mpg_melted, x='Transmission Type', y='Average MPG', hue='MPG Type', palette='Set2')
plt.title('Average MPG (Highway vs City) by Transmission Type', fontsize=14)
plt.xlabel('Transmission Type')
plt.ylabel('Average MPG')
plt.legend(title='MPG Type')
plt.tight_layout()
plt.show()
```
