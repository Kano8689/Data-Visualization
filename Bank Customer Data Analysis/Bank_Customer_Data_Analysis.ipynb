# **A. Distributions**

### **1. Plot distribution of age using histogram.**

```python
sns.histplot(data=data, x="age", bins = 10)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Age")
plt.show()
```

### **2. Create boxplot for account balance.**

```python
boxPlot = px.box(data,
                 x="balance",
                 labels={"balance":"Account Balance"},
                 title="Box Plot: Account Balance")
boxPlot.show()
```

### **3. Plot countplot for job categories.**

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=data, x="job")
plt.xticks(rotation=45)
plt.xlabel("Job Categories")
plt.ylabel("Employee Count")
plt.title("Countplot for Job Categories")
plt.show()
```

### **4. Visualize marital status distribution.**

```python
histData = data.sort_values(by="age")
maritalHist = px.histogram(histData,
                            x="marital",
                            labels={"marital":"Marital Status"},
                            title="Marital Status Distribution",
                            animation_frame="age")
maritalHist.show()
```

# **C. Campaign Analysis**

### **9. Plot campaign contacts distribution.**

```python
cmpnCntc = px.bar(data,
                  x="contact",
                  y="campaign",
                  labels={"campaign":"Campaign Contacts","contact":"Contact"},
                  title="Campaign Contacts Distribution")
cmpnCntc.show()
```

### **10. Analyze month-wise marketing campaigns.**

```python
mnthCmpnData = data
months = ["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
mnthCmpnData['month'] = pd.Categorical(mnthCmpnData['month'], categories=months, ordered=True)
mnthCmpnData = mnthCmpnData.sort_values(["day", "month"])

mnthCmpn = px.box(mnthCmpnData,
                  x="month",
                  y="campaign",
                  animation_frame="day",
                  labels={"month":"Month", "campaign":"Campaign Count"},
                  title="Month-wise Marketing Campaigns")
mnthCmpn.show()
```

### **11. Create bar chart of previous campaign outcomes.**

```python
prvsCmpn = px.bar(
    data,
    x="poutcome",
    y="previous",
    color="poutcome",
    title="Previous Campaign Outcomes",
    color_discrete_sequence=px.colors.qualitative.Bold
)

prvsCmpn.show()
```

# **G. Analysis Questions**

### **1. Which age group has the highest account balance?**

```python
data1 = data
bins = [18,30,40,50,60,70,80,90,100]
labels = ["18-30","31-40","41-50","51-60","61-70","71-80","81-90","91+"]
data1["age_group"] = pd.cut(data1["age"], bins=bins, labels=labels)
age_balance = data1.groupby("age_group")["balance"].mean().reset_index()
sns.barplot(x="age_group", y="balance", data=age_balance)
plt.show()
```

**Answer:**

- In above chart we can clearly see *Age* Group *81-90* has *High Bank Balance.*

### **2. Which job category subscribes the most to deposits?**

**Answer:**

  - There is defined *Multiple Bar Chart* *Job Category vs Deposit Subscription* In figure 14th. There we can clearly see *Management* has *High Deposit* as well as it has *High No-Deposit.*

### **3. Which month has the highest marketing activity?**

```python
monthData = data["month"].value_counts().reset_index()
monthData.columns = ["month", "count"]
sns.barplot(x="month", y="count", data=monthData)
plt.show()
```

**Answer:**

  - In above bar chart we can clearly guess that *May* month has *Highest Marketing.*

### **4. Does longer call duration increase deposit subscription?**

```python
sns.boxplot(x="deposit", y="duration", data=data)
plt.show()
```

**Answer:**

  - In above boxplot we can see *Deposit-Yes* is *Higher* Than *Deposit-No* so it's conclusion is *Yes, Longer call duration increases deposit subscription.*

### **5. Which education level responds best to campaigns?**

```python
eduYs = data[data["deposit"] == "yes"]
eduCnts = eduYs["education"].value_counts().reset_index()
eduCnts.columns = ["education", "count"]
sns.barplot(x="education", y="count", data=eduCnts)
plt.show()
```

**Answer:**

  - Above bar chart clearly shown that *Secondary Education Level* responds *Best* to campaigns.
